# 1. image classification With YOLO (free version)

Can classify: 

- **human**: [person]
- **vehicles**: [bicycle, car, motorbike..]
- **animal**: [bird, cat, dog...]
- **sport_lifestyle**: [backpack, umbrella, handbag, tie...]
- **kitchen_stuff**: [bottle, wineglass...]
- **food**: [banana, apple...]
- **household**: [chair, sofa, pottedpland, bed...]1
- **misc**: [everything else...]

```python
import os
from pathlib import Path

from cv_utils import ImageProcessor
from cv_utils.detection import YOLOPipeline
from cv_utils.video import VideoProcessor
```

```python
from dotenv import load_dotenv

load_dotenv()
PROJECT_ROOT = Path(os.getenv("PROJECT_ROOT"))
DATA_DIR = Path(os.getenv("DATA_DIR"))
YOLO_DIR = DATA_DIR / "YOLO4"
```

```python
os.listdir(YOLO_DIR)
```

```python
labels_file = YOLO_DIR / "coco.names"
weights = YOLO_DIR / "yolov4.weights"
config = YOLO_DIR / "yolov4.cfg"

print("labels exist", labels_file.exists())
print("weights exist", weights.exists())
print("config exist", config.exists())
```

```python
processor = ImageProcessor(YOLO_DIR / "kitchen.png")
processor.show()
```

```python
yolo_pipe = YOLOPipeline(
    config_file=config,
    labels_file=labels_file,
    weights_file=weights,
    confidence_thres=0.1,
    score_thres=0.9,
    crops_path=YOLO_DIR / "output"
)
processor.pipeline = yolo_pipe

processed_img = processor.process()
```

```python
processor.img_path = YOLO_DIR / "skate.jpg"
processed_img = processor.process()
```
